# FOR CLASSIFICATION

```python
import numpy as np
from collections import Counter

class KNN:
    def __init__(self, k, distance=None):
        self.k = k
        self.distance = distance if distance is not None else self.default_distance

    def train(self, x_train, y_train):
        self.x_train = x_train
        self.y_train = y_train

    def predict(self, x_test):
        pred = [self.predict_cal(x_test_point) for x_test_point in x_test]
        return pred

    def predict_cal(self, x_test_point):
        # Calculate distances for each train point
        distances = [self.distance(x_test_point, x_train_point) for x_train_point in self.x_train]

        # Get indices of k smallest distances
        labels = np.argsort(distances)[:self.k]

        # Get the labels corresponding to the closest points
        selected_points = [self.y_train[index] for index in labels]

        # Count the votes and return the most common one
        majority_votes = Counter(selected_points).most_common(1)

        # Return the predicted label (not the count)
        return majority_votes[0][0] # (label , count)  # i want label(prediction not count)

    def default_distance(self, x1, x2):
        # Default Euclidean distance calculation
        return np.sqrt(np.sum((x1 - x2) ** 2))



```

# FOR REGRESSION

```python
import numpy as np

class KnnRegression:
    def __init__(self, k, distance="euclidean", p=3):
        self.k = k
        self.distance_type = distance
        self.p = p  # for Minkowski

    def fit(self, x_train, y_train):
        self.x_train = np.array(x_train)
        self.y_train = np.array(y_train)

    def predict(self, x_test):
        return [self._cal(x_test_point) for x_test_point in x_test]

    def _cal(self, x_test_point):
        distances = [self._compute_distance(x_test_point, x_train_point)
                     for x_train_point in self.x_train]
        index = np.argsort(distances)[:self.k]
        selected_points = self.y_train[index]
        return np.mean(selected_points)

    def _compute_distance(self, x1, x2):
        if self.distance_type == "euclidean":
            return np.sqrt(np.sum((x1 - x2)**2))
        elif self.distance_type == "manhattan":
            return np.sum(np.abs(x1 - x2))
        elif self.distance_type == "minkowski":
            return (np.sum(np.abs(x1 - x2)**self.p))**(1 / round((self.p),3))
        else:
            raise ValueError("Unsupported distance type")
```
