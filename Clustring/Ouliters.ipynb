# OUTLIERS
- An outlier is a data point that is significantly different from the rest of the observations in a dataset.

- 👉 In simple words:
It’s a value that is unusually high or low compared to most of the other data points.

# Why are outliers dangerous?

Outliers can hurt your model in several ways:

## Distorting statistics

The mean shifts heavily toward the outlier, which may misrepresent the data.

Example: Salaries [50k, 52k, 55k, 60k, 10M] → average salary looks very high because of one person.

## Messing with assumptions

Many models (like linear regression, logistic regression, PCA) assume that data roughly follows a normal distribution. Outliers break this assumption.

## Hurting model performance

In regression: a single outlier can pull the regression line up/down, reducing accuracy.

In clustering: an outlier may form its own useless cluster.

In distance-based models (like KNN): outliers distort distance calculations.

## Causing overfitting

The model may try too hard to "explain" the rare outlier, instead of learning the general trend.

#** Outlier Detection using the Percentile Method | Winsorization Technique**

- percentile tells you the value below which a certain percentage of data falls.

- 👉 Example: If your score is at the 90th percentile, it means you scored better than 90% of the people.

**Outlier Detection using Percentile Method**

- In this method, we use percentiles (like 1st, 5th, 95th, 99th) to identify extreme values.
- The percentiles we use should be symetrical means if upper limit is 95 percentile then lower should be 5 , if upper is 99 then lower should be 1

- Anything below the lower percentile (e.g., 1st or 5th) or above the upper percentile (e.g., 95th or 99th) is treated as a potential outlier.

- 👉 Example:
If the 5th percentile of income data = $10,000 and the 95th percentile = $80,   000, then:

- Any value < 10,000 or   >   80,000 is considered an outlier.

IF capping is done using percentiles then this is called winsorization

**Winsorization Technique**

- Instead of removing outliers, Winsorization replaces them with the closest boundary percentile.

- This reduces the impact of extreme values while keeping all data points.

- 👉 Example:

- If a value = $200,000 but the 95th percentile = $80,000, then Winsorization will replace 200,000 → 80,000.

- Similarly, very small values are pushed up to the lower boundary (e.g., below 5th percentile → replaced with 5th percentile).
